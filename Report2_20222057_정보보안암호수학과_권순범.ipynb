```python
def input_matrix(n):
    matrix = []
    print("5×5 관계행렬을 행 단위로 입력하세요.")
    print("(각 행은 0과 1을 공백으로 구분하여 5개씩 입력)")

    for _ in range(n):
        row = list(map(int, input().split()))
        while len(row) != n:
            print("오류: 정확히 5개의 0/1을 입력해야 합니다. 다시 입력하세요.")
            row = list(map(int, input().split()))
        matrix.append(row)
    return matrix


def print_matrix(matrix):
    for row in matrix:
        print(" ".join(map(str, row)))
    print()

def print_relation(matrix, A):
    size = len(matrix)
    for i in range(size):
        for j in range(size):
            if matrix[i][j] == 1:
                print(f"({A[i]}, {A[j]})", end=' ')
    print()

def print_paths_length_k(matrix, A, length):
    n = len(matrix)
    paths = []

    def dfs(current, path, depth):
        if depth == length:
            paths.append(path[:])
            return
        for nxt in range(n):
            if matrix[current][nxt] == 1:
                path.append(nxt)
                dfs(nxt, path, depth + 1)
                path.pop()

    for start in range(n):
        dfs(start, [start], 0)

    print(f"\n길이가 {length}인 모든 경로:")
    if not paths:
        print(f"→ 길이가 {length}인 경로가 없음")
    else:
        for path in paths:
            verts = [A[i] for i in path]
            print(" -> ".join(map(str, verts)),
                  f"=> ({verts[0]}, {verts[-1]})")

def is_reflexive(matrix):
    size = len(matrix)
    for i in range(size):
        if matrix[i][i] != 1:
            return False
    return True

def is_symmetric(matrix):
    size = len(matrix)
    for i in range(size):
        for j in range(size):
            if matrix[i][j] != matrix[j][i]:
                return False
    return True

def matrix_mult(A, B):
    size = len(A)
    result = [[0] * size for _ in range(size)]
    for i in range(size):
        for j in range(size):
            for k in range(size):
                result[i][j] |= A[i][k] & B[k][j]
    return result

def is_transitive(matrix):
    size = len(matrix)
    power = [row[:] for row in matrix]
    combined = [row[:] for row in matrix]
    for _ in range(1, size):
        power = matrix_mult(power, matrix)
        for i in range(size):
            for j in range(size):
                combined[i][j] |= power[i][j]
    return combined == matrix

def get_equivalence_class(matrix, A, idx):
    eq_class = []
    for j in range(len(A)):
        if matrix[idx][j] == 1:
            eq_class.append(A[j])
    return eq_class

def print_equivalence_classes(matrix, A):
    print("\n각 원소의 동치류:")
    for i in range(len(A)):
        eq = get_equivalence_class(matrix, A, i)
        print(f"[{A[i]}] = {{{', '.join(map(str, eq))}}}")

def reflexive_closure(matrix):
    size = len(matrix)
    result = [row[:] for row in matrix]
    for i in range(size):
        result[i][i] = 1
    return result


def symmetric_closure(matrix):
    size = len(matrix)
    result = [row[:] for row in matrix]
    for i in range(size):
        for j in range(size):
            if result[i][j] == 1 or result[j][i] == 1:
                result[i][j] = 1
                result[j][i] = 1
    return result


def transitive_closure(matrix):
    size = len(matrix)
    result = [row[:] for row in matrix]
    for k in range(size):
        for i in range(size):
            for j in range(size):
                if result[i][k] and result[k][j]:
                    result[i][j] = 1
    return result



def main():
    A = [1, 2, 3, 4, 5]
    n = 5

    # 기능 1
    matrix = input_matrix(n)
    print("\n입력된 관계행렬:")
    print_matrix(matrix)
    print("관계 R에 포함된 순서쌍:")
    print_relation(matrix, A)

    # 추가기능
    print_paths_length_k(matrix, A, length=3)

    # 기능 2 + 기능 3
    print("\n[반사성]")
    print("Yes" if is_reflexive(matrix) else "No")

    print("\n[대칭성]")
    print("Yes" if is_symmetric(matrix) else "No")

    print("\n[추이성]")
    print("Yes" if is_transitive(matrix) else "No")

    if is_reflexive(matrix) and is_symmetric(matrix) and is_transitive(matrix):
        print("\n⇒ 동치 관계이므로 동치류 출력:")
        print_equivalence_classes(matrix, A)
    else:
        print("\n⇒ 동치 관계가 아님")

    # 기능 4
    print("\n=== 반사 폐포 ===")
    print("변환 전 행렬:")
    print_matrix(matrix)
    ref = reflexive_closure(matrix)
    print("변환 후 행렬(반사 폐포):")
    print_matrix(ref)

    print("\n=== 대칭 폐포 ===")
    print("변환 전 행렬:")
    print_matrix(matrix)
    sym = symmetric_closure(matrix)
    print("변환 후 행렬(대칭 폐포):")
    print_matrix(sym)

    print("\n=== 추이 폐포 ===")
    print("변환 전 행렬:")
    print_matrix(matrix)
    trn = transitive_closure(matrix)
    print("변환 후 행렬(추이 폐포):")
    print_matrix(trn)


    print("\n=== 동치관계 재판별 ===")
    matrix = reflexive_closure(matrix)
    matrix = symmetric_closure(matrix)
    matrix = transitive_closure(matrix)
    print("변환 후 행렬:")
    print_matrix(matrix)
    if is_reflexive(matrix) and is_symmetric(matrix) and is_transitive(matrix):
        print("\n⇒ 동치 관계이므로 동치류 출력:")
        print_equivalence_classes(matrix, A)
    else:
        print("\n⇒ 동치 관계가 아님")


# 실행
if __name__ == "__main__":
    main()
```
